<h1>Bibliotecas</h1>

```python
import requests
import pandas as pd

from datetime import datetime
from pathlib import Path
```

<h1>Camara dos Deputados</h1>

<h2>Consulta - API</h2>

```python
URL = 'https://dadosabertos.camara.leg.br/api/v2/proposicoes'

params = {
    "ano": 2026,
    "itens": 20,
    "ordem": "DESC",
    "ordenarPor": "id"
}

resposta = requests.get(URL, params=params)
print(resposta.status_code)

dados = resposta.json()

proposicoes = dados.get('dados', [])

df_raw = pd.DataFrame(proposicoes)

df_raw.head()
```

<h2>Extração</h2>

```python
data_agora = datetime.now().strftime('%Y-%m-%d_%H-%M-%S')

caminho_saida = f'../dados/pipeline_verdadeiro_fontes_oficiais/raw/camara_proposicoes_raw_{data_agora}.csv'

df_raw.to_csv(caminho_saida, index=False, encoding='utf-8-sig')

print(f'Arquivo bruto salvo em: {caminho_saida}')
```

<h2>Curated</h2>

```python
df_curated = df_raw.copy()

df_curated['texto_afirmacao'] = (

    'A proposição '
    + df_curated['siglaTipo'].astype(str)
    + ' '
    + df_curated['numero'].astype(str)
    + '/'
    + df_curated['ano'].astype(str)
    + 'foi registrada na Câmara Dos Deputados.'
)

df_curated['rotulo_binario'] = 'VERDADEIRO'
df_curated['fonte_verificada'] = 'CAMARA_DOS_DEPUTADOS'
df_curated['subtipo_falsidade'] = 'NA'
df_curated['tema'] = 'CONGRESSO'
df_curated['status_revisao'] = 'PENDENTE'

df_curated = df_curated[
    [
        'id',
        'texto_afirmacao',
        'rotulo_binario',
        'subtipo_falsidade',
        'tema',
        'fonte_verificada',
        'uri',
        'status_revisao'

    ]
].rename(columns={'uri':'url_fonte'})

df_curated.head()
```
